```python
import numpy as np 
# import sklearn as sk
import matplotlib.pyplot as plt
import pandas as pd
import sympy as sp
```

```python
def pmatrix(a, digit : int =3, T : bool =False):
    """Returns a LaTeX bmatrix
    :a: numpy array
    :returns: LaTeX bmatrix as a string
    """
    if len(a.shape) > 2:
        raise ValueError('bmatrix can at most display two dimensions')
    lines = str(np.round(a, digit)).replace('[', '').replace(']', '').splitlines()
    rv = [r'\begin{pmatrix}']
    if not T:
        rv += ['  ' + ' & '.join(l.split()) + r'\\' for l in lines]
    else:
        rv += ['  ' + '\\\\ '.join(l.split()) for l in lines]
    rv +=  [r'\end{pmatrix}']
    return '\n'.join(rv)
```

```python
x1 = np.array([-1, -.8, -.6, -.4, -.2, 0, 0.2, .4, .6, 1])
x2 = np.array([-.2, -.8, 1, .4, -.6, 0.2, .6, 0, .8, -1])
y = np.array([-6.805, -7.433, -2.080, -7.106, -3.465, -3.103, -2.177, -1.333, 0.091, -0.282])


y_pi = np.array([-1.622, -1.428, -1.373, -1.468, -1.526, -1.597, -1.542, -1.625, -1.697, -1.551, 
             -1.675, -1.567, -1.587, -1.507, -1.655, -1.550, -1.667, -1.601, -1.444, -1.517])
```

```python
F = np.array([[1 for _ in range(10)],
    x1, 
    x2, 
    x1**2,
    x1*x2,
    x2**2
])
```

```python
F.T # Матрица регрессоров
```

```python
print(pmatrix(F.T))
print(pmatrix(y, T=True))
```

```python
vrn = F @ y 
print(pmatrix(vrn, T=True)) # вектор системы нормальных уравнений
vrn
```

```python
G = F @ F.T
print(pmatrix(G)) # информационная матрица
```

```python
C = np.linalg.inv(F@F.T)
print(pmatrix(np.round(C, 3))) # матрица дисперсий-ковариаций
```

```python
b = C @ vrn
print(np.round(b, 3)) # вектор оценок регрессионых
# коэффициентов без учета заданных линейных ограничений
print(pmatrix(b, T=True))
```

```python
def lin_model(b):
    return np.round(b[0] + b[1] * x1 + b[2] * x2 + b[3] * x1**2 + b[4] * x1*x2 + b[5] * x2**2, 3)
```

```python
y_p_mean = y_pi.mean()
print(y_p_mean)
```

```python
# находим оценку дисперсии
s_pow = np.sum((y_pi - y_p_mean)**2) / 19
print("Оценка дисперсии:", round(s_pow, 5))
```

```python
# ковариационная матрица оценок V[b]

print(pmatrix(C * s_pow, 4))
```

```python
fx = F.T[6] # (0.2, 0.6)
print(pmatrix(fx))
print("Оценка дисперсии предсказанного значения отклика $s^2[y(x)] = ", np.round(fx @ C @ fx.T * s_pow, 4), "$")
```
